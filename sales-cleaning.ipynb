```python
import pandas as pd
```

```python
df = pd.read_csv(r'amazon_CLEANED.csv', encoding='utf-8-sig')
print(f"Loaded: {df.shape[0]} rows, {df.shape[1]} columns")
print(df)
```

```python
#fix1 Re-encode text columns cleanly
# Strip out non-UTF-8 characters that corrupt the import
text_cols = [
    'product_name', 'about_product', 'review_title',
    'review_content', 'user_name', 'category'
]
for col in text_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.encode('latin-1', errors='ignore')
        .str.decode('utf-8', errors='ignore')
    )

#FIX 2: Clean discount_percentage — remove % sign
df['discount_percentage'] = (
    df['discount_percentage']
    .astype(str)
    .str.replace('%', '', regex=False)
    .str.strip()
    .astype(float)
)
print("discount_percentage sample:", df['discount_percentage'].head(3).tolist())

#FIX 3: Confirm is_high_rated is boolean
df['is_high_rated'] = df['is_high_rated'].astype(bool)
print("is_high_rated sample:", df['is_high_rated'].head(3).tolist())

#FIX 4: Strip leading/trailing whitespace from all text
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip()

#Save as clean UTF-8 CSV
output_path = r'C:\Users\hp\amazon-sales-analysis\amazon_READY.csv'
df.to_csv(output_path, index=False, encoding='utf-8-sig')

print(f"\nDone. Clean file saved to: {output_path}")
print(f"Rows: {df.shape[0]} | Columns: {df.shape[1]}")
```
